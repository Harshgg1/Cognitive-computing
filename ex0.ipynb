```python
print("Name: Harsh")
print("Roll No: 1024170219")

def prime(n):
    if n<2:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
    return True

n=int(input("Enter Number: "))

if prime(n):
    print("Prime")
else:
    print("Not Prime")
```
