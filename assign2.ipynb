```python
roll_no = 1024170219

L = []

for digit in str(roll_no):
    L.append(int(digit) * 10)

print("L =", L)

L.append(50)
print("After append:", L)

L.insert(2, 60)
print("After insert:", L)

L.remove(60)
print("After remove:", L)

L.pop()
print("After pop:", L)

L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)

print("First 3 and last 3:", L[:3], L[-3:])

average = sum(L) / len(L)

new_list = [x for x in L if x > average]

print("Average =", average)
print("Elements greater than average:", new_list)
```

```python
scores = (10, 0, 20, 40, 10, 70, 0, 20)

highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Lowest score appears:", lowest_count, "times")



reversed_scores = list(scores[::-1])

print("Reversed tuple as list:", reversed_scores)

# tuple cannot be changed in place because tuples are immutable.


score = int(input("Enter a score: "))

if score in scores:
    print("First occurrence index:", scores.index(score))
else:
    print("Score not present")


try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)

# tuples are immutable so their elements cannot be changed directly
# lists are mutable so list elements can be changed



first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
# Q3

import random

random.seed(1024170219)

numbers = []

for i in range(100):
    number = random.randint(100, 900)
    numbers.append(number)

print("Random numbers:")
print(numbers)



odd_count = 0

for number in numbers:
    if number % 2 != 0:
        odd_count += 1

print("Number of odd numbers:", odd_count)



even_count = 0

for number in numbers:
    if number % 2 == 0:
        even_count += 1

print("Number of even numbers:", even_count)



prime_numbers = []

for number in numbers:

    is_prime = True

    if number < 2:
        is_prime = False

    else:
        for i in range(2, number):
            if number % i == 0:
                is_prime = False
                break

    if is_prime:
        prime_numbers.append(number)

print("Number of prime numbers:", len(prime_numbers))
print("Prime numbers:", prime_numbers)



most_frequent = numbers[0]
highest_count = numbers.count(numbers[0])

for number in numbers:

    count = numbers.count(number)

    if count > highest_count:
        highest_count = count
        most_frequent = number

print("Most frequent number:", most_frequent)
print("It occurs:", highest_count, "times")
```

```python
roll_no = 1024170219

digits = []

for digit in str(roll_no):
    digits.append(int(digit))

A = set()
B = set()

for digit in digits:
    A.add(digit * 7)
    B.add(digit * 9)

print("A =", A)
print("B =", B)


print("Union:", A.union(B))


print("Intersection:", A.intersection(B))



print("A - B:", A.difference(B))
print("B - A:", B.difference(A))

# difference() gives the elements present in one set
# but not present in the other set.



print("Symmetric difference:", A.symmetric_difference(B))

# it gives elements that are in A or B,
# but not in both.


print("Is A subset of B?", A.issubset(B))
print("Is B superset of A?", B.issuperset(A))



x = int(input("Enter a value to remove from A: "))

A.discard(x)

print("A after discard:", A)

# discard() does not give an error if the value is not present.
# remove() would give an error if the value is not present.
```

```python
my_dict = {
    "name": "Harsh",
    "roll_no": "1024170219",
    "branch": "Computer Engineering",
    "age": 20,
    "city": "Punjab"
}

print("Original dictionary:")
print(my_dict)



my_dict["location"] = my_dict.pop("city")

print("After renaming city to location:")
print(my_dict)

# pop() removes city and returns its value,
# which is then stored in location.



my_dict["cgpa"] = 8.77

print("After adding CGPA:")
print(my_dict)



my_dict["age"] = my_dict["age"] + 1

print("After increasing age:")
print(my_dict)



dict1 = my_dict.copy()

dict1.pop("branch")

print("After deleting branch using pop():")
print(dict1)



dict2 = my_dict.copy()

del dict2["branch"]

print("After deleting branch using del:")
print(dict2)

# pop() returns the value that it removes,
# while del simply removes the key-value pair.



print("Key-value pairs:")

for key, value in my_dict.items():
    print(key, "->", value)



if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key not found")



friend_dict = {
    "name": "Rahul",
    "roll_no": "1024170220",
    "branch": "Computer Engineering",
    "age": 20,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("Merged dictionary:")
print(merged_dict)

# If both dictionaries have the same key,
# the value from friend_dict will replace the value from my_dict.



string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("Only string values:")
print(string_values)
```
