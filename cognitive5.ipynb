```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)
print("Addition of 2:", arr + 2)
print("Multiplication by 3:", arr * 3)
print("Division by 2:", arr / 2)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

max_count = counts.max()
most_frequent = values[counts == max_count]

print("Most frequent value:", most_frequent)
print("Frequency:", max_count)

for value in most_frequent:
    print("Indices of", value, ":", np.where(x == value)[0])
```

```python
import numpy as np

harsh_garg = np.linspace(10, 100, 25)

print("Array:", harsh_garg)
print("Dimensions:", harsh_garg.ndim)
print("Shape:", harsh_garg.shape)
print("Total elements:", harsh_garg.size)
print("Data type:", harsh_garg.dtype)
print("Total bytes:", harsh_garg.nbytes)

transpose = harsh_garg.reshape(25, 1)

print("Transpose using reshape():")
print(transpose)

print("Transpose using T:")
print(harsh_garg.T)
```

```python
import numpy as np

ucs420_harsh_garg = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Array:")
print(ucs420_harsh_garg)

print("Mean:", np.mean(ucs420_harsh_garg))
print("Median:", np.median(ucs420_harsh_garg))
print("Maximum:", np.max(ucs420_harsh_garg))
print("Minimum:", np.min(ucs420_harsh_garg))
print("Unique elements:", np.unique(ucs420_harsh_garg))

reshaped_ucs420_harsh_garg = ucs420_harsh_garg.reshape(4, 3)

print("Reshaped array:")
print(reshaped_ucs420_harsh_garg)

resized_ucs420_harsh_garg = np.resize(ucs420_harsh_garg, (2, 3))

print("Resized array:")
print(resized_ucs420_harsh_garg)
```
