```python
import pandas as pd

roll_no = "1024170219"

fixed_entries = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.", "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.", "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.", "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.", "keywords": "pay payment upi fee", "category": "billing"}
]

personalized_entries = [
    {"question": "how do i update my mobile number", "answer": "Go to Account Settings and update your registered mobile number.", "keywords": "mobile number update", "category": "account"},
    {"question": "how can i check my payment status", "answer": "Go to the payment section to check your payment status.", "keywords": "payment status billing", "category": "billing"}
]

df = pd.DataFrame(fixed_entries + personalized_entries)

print(df)
```

```python
def score_query(query, df):
    words = query.lower().split()
    results = []

    for _, row in df.iterrows():
        keywords = row["keywords"].lower().split()
        score = sum(word in keywords for word in words)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "confidence": score
            })

    return sorted(results, key=lambda x: x["confidence"], reverse=True)

query = input("Enter your question: ")

results = score_query(query, df)

for result in results:
    print(result)
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]

print(same_category("account", df))
```

```python
def score_query_with_ties(query, df):
    results = score_query(query, df)

    if not results:
        print("No matching entries found.")
        return

    highest_score = results[0]["confidence"]

    best_matches = [
        result for result in results
        if result["confidence"] == highest_score
    ]

    if len(best_matches) > 1:
        print("Tie found:")
        for result in best_matches:
            print(result)
    else:
        print("Best match:")
        print(best_matches[0])

print("Tie example:")
score_query_with_ties("fee", df)

print("\nNon-tie example:")
score_query_with_ties("password", df)
```
