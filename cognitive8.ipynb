```python
text = """
Technology has changed the way people learn and communicate.
I enjoy exploring computer science because it combines logic and creativity.
Programming allows me to build useful applications and solve real problems.
Artificial intelligence helps computers understand data and make predictions.
Learning new technologies improves my skills and prepares me for the future.
I believe technology can make education more accessible and efficient.
"""
print(text)
```

```python
import re

clean_text = re.sub(r'[^\w\s]', '', text.lower())
print(clean_text)
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

words = nltk.word_tokenize(clean_text)
sentences = nltk.sent_tokenize(text)

print("Words:", words)
print("\nSentences:", sentences)
```

```python
split_words = text.split()
tokenized_words = nltk.word_tokenize(text)

print("Using split():", split_words)
print("\nUsing word_tokenize():", tokenized_words)
```

```python
nltk.download('stopwords')

from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word.isalpha() and word not in stop_words
]

print("Words after removing stopwords:")
print(filtered_words)
```

```python
from nltk import FreqDist

fdist = FreqDist(filtered_words)

print("Word Frequencies:")
for word, frequency in fdist.most_common():
    print(word, ":", frequency)

fdist.plot(10, title="Top Word Frequencies")
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("Stemmed words:")
print(stemmed_words)
```

```python
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word) for word in filtered_words
]

print("Lemmatized words:")
print(lemmatized_words)
```

```python
documents = [
    "Artificial intelligence helps computers learn from data.",
    "Machine learning improves predictions using data.",
    "Cloud computing provides scalable computing resources."
]
```

```python
from sklearn.feature_extraction.text import CountVectorizer

count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(documents)

print("Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

print("TF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

```python
feature_names = tfidf_vectorizer.get_feature_names_out()
scores = tfidf_matrix.toarray()

for i, document in enumerate(documents):
    print(f"\nDocument {i + 1}:")

    top_indices = scores[i].argsort()[::-1][:3]

    for index in top_indices:
        print(feature_names[index], ":", round(scores[i][index], 3))
```

```python
text1 = """
Artificial intelligence enables machines to learn from data.
It helps automate tasks and make intelligent predictions.
AI is used in healthcare, education, and transportation.
Machine learning is an important part of artificial intelligence.
"""

text2 = """
Blockchain stores information in a decentralized digital ledger.
It improves transparency and security in digital transactions.
Blockchain is used in finance, supply chains, and digital identity.
Cryptography helps protect blockchain transactions.
"""
```

```python
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    tokens = text.split()
    return tokens

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("Text 1 tokens:", tokens1)
print("\nText 2 tokens:", tokens2)
```

```python
set1 = set(tokens1)
set2 = set(tokens2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard_similarity = len(intersection) / len(union)

print("Common words:", intersection)
print("Jaccard Similarity:", round(jaccard_similarity, 4))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer()
tfidf_vectors = vectorizer.fit_transform([text1, text2])

cosine_score = cosine_similarity(tfidf_vectors[0], tfidf_vectors[1])[0][0]

print("Cosine Similarity:", round(cosine_score, 4))
```

```python
print("Jaccard Similarity:", round(jaccard_similarity, 4))
print("Cosine Similarity:", round(cosine_score, 4))

if cosine_score > jaccard_similarity:
    print("Cosine similarity gives a higher similarity score.")
elif jaccard_similarity > cosine_score:
    print("Jaccard similarity gives a higher similarity score.")
else:
    print("Both metrics give the same similarity score.")
```

```python
reviews = [
    "This laptop is amazing and performs very well.",
    "The product is excellent and easy to use.",
    "The battery is terrible and the performance is poor.",
    "The product is okay but nothing special."
]
```

```python
from textblob import TextBlob

for review in reviews:
    sentiment = TextBlob(review).sentiment

    print("\nReview:", review)
    print("Polarity:", round(sentiment.polarity, 3))
    print("Subjectivity:", round(sentiment.subjectivity, 3))
```

```python
def classify_sentiment(review):
    polarity = TextBlob(review).sentiment.polarity

    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"

positive_reviews = []

for review in reviews:
    category = classify_sentiment(review)
    print(category, ":", review)

    if category == "Positive":
        positive_reviews.append(review)
```

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud

if positive_reviews:
    positive_text = " ".join(positive_reviews)

    wordcloud = WordCloud(
        width=800,
        height=400,
        background_color="white"
    ).generate(positive_text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Word Cloud of Positive Reviews")
    plt.show()
else:
    print("No positive reviews found.")
```

```python
training_text = """
Technology has become an important part of modern life.
Computers help students learn new concepts and solve difficult problems.
Artificial intelligence allows machines to recognize patterns and make predictions.
Programming teaches logical thinking and helps developers create useful applications.
Data science helps people analyze information and discover meaningful insights.
Cloud computing makes it possible to store data and run applications online.
Cybersecurity protects computer systems from unauthorized access and digital threats.
Machine learning improves when models learn from relevant training data.
Students can develop technical skills by building projects and practicing regularly.
Technology continues to create opportunities in education, business, healthcare, and research.
Learning technology requires patience, creativity, and consistent effort.
Future innovations will help people solve complex problems more efficiently.
"""
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer()
tokenizer.fit_on_texts([training_text])

total_words = len(tokenizer.word_index) + 1

print("Vocabulary size:", total_words)
print("Word index:", tokenizer.word_index)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

input_sequences = []

for sentence in training_text.lower().split("."):
    sequence = tokenizer.texts_to_sequences([sentence.strip()])[0]

    for i in range(2, len(sequence) + 1):
        input_sequences.append(sequence[:i])

max_sequence_len = max(len(seq) for seq in input_sequences)

padded_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding="pre"
)

X = padded_sequences[:, :-1]
y = padded_sequences[:, -1]

print("Input shape:", X.shape)
print("Output shape:", y.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

model = Sequential([
    Embedding(total_words, 32, input_length=max_sequence_len - 1),
    LSTM(64),
    Dense(total_words, activation="softmax")
])

model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.fit(X, y, epochs=150, verbose=1)
```

```python
def generate_text(seed_text, next_words):
    result = seed_text

    for _ in range(next_words):
        sequence = tokenizer.texts_to_sequences([result])[0]

        sequence = pad_sequences(
            [sequence],
            maxlen=max_sequence_len - 1,
            padding="pre"
        )

        predicted_probs = model.predict(sequence, verbose=0)[0]
        predicted_word_index = np.argmax(predicted_probs)

        output_word = tokenizer.index_word.get(predicted_word_index, "")

        if not output_word:
            break

        result += " " + output_word

    return result

seed_text = "technology helps"

for i in range(3):
    print(generate_text(seed_text, 8) + ".")
```
