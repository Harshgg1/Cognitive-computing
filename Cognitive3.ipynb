```python
import pandas as pd

# Create dataset
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Harsh", "nikhil", "kiwi", "pamma", "bidi"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(data)

# a) Shape
print("Shape:", employees.shape)

# b) Summary
employees.info()

# c) Descriptive statistics
print(employees.describe())

# d) First 5 and last 3 rows
print(employees.head(5))
print(employees.tail(3))

# e) Statistics
print("Average Salary:", employees["Salary"].mean())
print("Total Bonus:", employees["Bonus"].sum())
print("Youngest Age:", employees["Age"].min())
print("Highest Rating:", employees["Rating"].max())

# f) Sort by salary descending
employees = employees.sort_values(by="Salary", ascending=False)
print(employees)

# g) Performance category
def performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance"] = employees["Rating"].apply(performance)

# h) Missing values
print(employees.isnull().sum())

# i) Rename column
employees = employees.rename(columns={"Employee_ID": "ID"})

# j) Employees with more than 5 years experience
print(employees[employees["Years_of_Experience"] > 5])

# Employees in IT
print(employees[employees["Department"] == "IT"])

# k) Add Tax
employees["Tax"] = employees["Salary"] * 0.10

print(employees)

# l) Save to CSV
employees.to_csv("modified_employees.csv", index=False)

print("Modified file saved successfully.")
```
